```python
import psycopg2
import pandas as pd
import yfinance as yf
import concurrent.futures
import time
from datetime import datetime

DB_CONFIG = {
    "dbname": "stocks",
    "user": "admin",
    "password": "password",
    "host": "localhost",
    "port": 5434
}

START_DATE = "2015-01-01"
YF_TIMEOUT = 10  # seconds for each yfinance request
RETRIES = 2      # retry twice before skipping


def db_connect():
    return psycopg2.connect(**DB_CONFIG)


def get_all_stocks(cur):
    cur.execute("SELECT id, symbol, exchange FROM stocks WHERE instrument = 'EQ' OFFSET 430;")
    rows = cur.fetchall()
    stock_list = []

    for stock_id, symbol, exchange in rows:
        if exchange == "NSE":
            symbol_yf = symbol + ".NS"
        elif exchange == "BSE":
            symbol_yf = symbol + ".BO"
        else:
            symbol_yf = symbol

        stock_list.append((stock_id, symbol, symbol_yf))

    return stock_list


def clean_yf_dataframe(df):
    # Flatten multiindex columns
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = ["_".join([str(c) for c in col if c]) for col in df.columns]
    else:
        df.columns = [str(c) for c in df.columns]

    # Standardize names
    rename_map = {}
    for col in df.columns:
        lc = col.lower()
        if "open" in lc and "open" not in rename_map:
            rename_map[col] = "Open"
        elif "high" in lc and "high" not in rename_map:
            rename_map[col] = "High"
        elif "low" in lc and "low" not in rename_map:
            rename_map[col] = "Low"
        elif "close" in lc and "adj" not in lc and "close" not in rename_map:
            rename_map[col] = "Close"
        elif "volume" in lc:
            rename_map[col] = "Volume"

    df = df.rename(columns=rename_map)

    required = ["Date", "Open", "High", "Low", "Close", "Volume"]
    if not all(col in df.columns for col in required):
        return None

    # Convert them properly
    for col in ["Open", "High", "Low", "Close", "Volume"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    return df[required]


def fetch_history(symbol_yf):
    """Fetch with timeout + retry + safe fallback."""
    def _download():
        return yf.download(
            symbol_yf,
            start=START_DATE,
            interval="1d",
            progress=False,
            auto_adjust=False
        )

    for attempt in range(RETRIES + 1):
        try:
            with concurrent.futures.ThreadPoolExecutor(max_workers=1) as executor:
                future = executor.submit(_download)
                df = future.result(timeout=YF_TIMEOUT)

            if df is None or df.empty:
                continue  # retry

            df = df.reset_index()
            df = clean_yf_dataframe(df)

            return df

        except concurrent.futures.TimeoutError:
            print(f"⏳ Timeout fetching {symbol_yf}, retry {attempt+1}/{RETRIES}")
        except Exception:
            pass

        time.sleep(1)  # small wait between retries

    return None  # failed after retries


def insert_history(cur, stock_id, df):
    rows = []

    for _, row in df.iterrows():
        rows.append((
            stock_id,
            pd.to_datetime(row["Date"]).date(),
            row["Open"],
            row["High"],
            row["Low"],
            row["Close"],
            0 if pd.isna(row["Volume"]) else int(row["Volume"])
        ))

    cur.executemany("""
        INSERT INTO historical_prices
        (stock_id, date, open, high, low, close, volume)
        VALUES (%s,%s,%s,%s,%s,%s,%s)
        ON CONFLICT (stock_id, date) DO NOTHING;
    """, rows)


def main():
    conn = db_connect()
    cur = conn.cursor()

    stocks = get_all_stocks(cur)
    print(f"Processing  {len(stocks)} EQ stocks...\n")

    for stock_id, symbol, symbol_yf in stocks:
        print(f"Fetching: {symbol} ({symbol_yf})")

        df = fetch_history(symbol_yf)

        if df is None:
            print(f"No data for {symbol}, skipping.\n")
            continue

        insert_history(cur, stock_id, df)
        conn.commit()

        print(f"Inserted {len(df)} rows for {symbol}\n")

    cur.close()
    conn.close()

    print("Completed processing all stocks.")


if __name__ == "__main__":
    main()
```
